# <center> Маршруты в графе

>**Маршрут в графе** — это путь, который начинается в одной вершине и следует по рёбрам к другим вершинам, не обязательно различным. Главное условие — между каждой парой последовательных вершин в маршруте должно существовать ребро. Иными словами, вы можете переходить от одной вершины к следующей, если они непосредственно соединены.

Важно отметить, что в маршруте вершины и рёбра могут повторяться. Это означает, что маршрут может проходить через одну и ту же вершину или ребро более одного раза.

## Начальные и конечные вершины

Маршрут позволяет описать путь через граф от одной точки к другой. Он определяется начальной и конечной вершинами.

>**Начальная вершина** — это точка, из которой маршрут начинает своё движение, а **конечная вершина** — это точка, в которой маршрут заканчивается.

Длина маршрута измеряется количеством входящих в него рёбер для  невзвешенного графа или суммой весов этих рёбер во взвешенном графе. Маршрут считается **замкнутым**, если начальная и конечная вершины совпадают, то есть путь возвращается в исходную точку. Это особый вид маршрута, который также известен как **цикл**.

## Открытый и закрытый маршрут

В теории графов существуют понятия «открытый маршрут» и «закрытый маршрут».

>**Открытый маршрут** — это такой маршрут, в котором начальная и конечная вершины не совпадают.

В открытом маршруте путь начинается в одной вершине и заканчивается в другой, не возвращаясь назад в начальную точку. 

>**Закрытый маршрут** — это маршрут, где начальная и конечная вершины совпадают.

Таким образом, если вы начали путь в вершине A, прошли через другие вершины и вернулись в вершину A, то маршрут считается закрытым.

Теперь, допустим, путешественник хочет посетить Золотое кольцо, начиная свой путь из Москвы. Он может двигаться к Сергиеву Посаду, затем к Переславлю-Залесскому, далее к Ростову Великому, и так далее, посещая каждый город по одному разу и проезжая по дорогам, которые связывают эти города. Тогда это будет уже особый тип маршрута — **цепь**, — в нём все вершины различны.



>**Цепь** можно описать как путь, проходящий через разные вершины, где каждое ребро пройдено только один раз.

Кроме того, что наш маршрут является цепью, мы можем обнаружить и ещё одно важное свойство. Так как у нас конечная и начальная вершины совпадают, то цепь является замкнутой и носит определённое название — **цикл**.

>**Циклы** — это замкнутые контуры, в которых можно вернуться в исходную точку, не проходя ни одну вершину более одного раза. Цикл длины 1 называется петлёй.

На карте обозначены расстояния между городами, и эти значения могут быть использованы для определения длины маршрута, если суммировать расстояния по пройденным дорогам. Так, маршрут Москва → Тверь → Великий Новгород имеет длину, равную сумме расстояний между Москвой и Тверью, а затем Тверью и Великим Новгородом.

## Связность

>**Связность** в теории графов — это характеристика, показывающая, насколько тесно связаны вершины графа между собой.

Это понятие принимает разные формы в зависимости от того, говорим ли мы о неориентированных или ориентированных графах, а также от того, какие именно свойства связности нас интересуют.

>Неориентированный граф считается **связным**, если существует путь между любой парой его вершин: например, отдельно взятый зелёный маршрут. Это значит, что вы можете добраться от любой вершины графа до любой другой, перемещаясь по рёбрам.

Если граф **не связный**, он содержит несколько изолированных подграфов, называемых компонентами связности.

Зелёный и розовый маршрут являются компонентами связности одного большого графа.

Маршруты, цепи и циклы имеют важное значение в различных областях, таких как компьютерные сети, логистика, биологические исследования и социальные сети. Они помогают анализировать и понимать сложные сетевые структуры, определять кратчайшие пути, выявлять потенциальные узлы перегрузки и анализировать циклические зависимости.

# <center> 3. Расстояния в графе

Мы уже говорили, что расстояние в графе определяется как длина кратчайшего пути между двумя вершинами, где длиной пути называется количество рёбер, которое он содержит в неориентированном графе, или сумма весов рёбер во взвешенном графе.

В следующих юнитах мы познакомим вас с ключевыми алгоритмами, которые используются для разных задач в графах:
* **Обход графа в ширину (BFS, Breadth-First Search)**, который применяется для нахождения кратчайшего пути по числу рёбер в невзвешенных графах и для определения уровней доступности от заданной начальной точки.
* **Обход графа в глубину (DFS, Depth-First Search)**, полезный для решения задач, требующих полного изучения графа, таких как поиск компонент связности или топологическая сортировка.
* **Алгоритм Дейкстры**, который является базовым методом для нахождения кратчайшего пути от одной вершины до всех остальных во взвешенном графе с неотрицательными весами рёбер.

Каждый из этих алгоритмов служит разным целям и оптимален в определённых контекстах, поэтому важно знать, когда и как их применять для эффективного решения задач на графах.

Каждый из них имеет свои преимущества и недостатки и выбирается в определённом случае. Но чтобы вы могли лучше понять эти алгоритмы, в этом юните мы разберём некоторые классические понятия.

## Эксцентриситет вершины в графе

>**Эксцентриситет вершины в графе** — это максимальное расстояние от этой вершины до любой другой вершины в графе.

Эта мера может быть использована для анализа сетевых структур: с её помощью можно идентифицировать критические точки или узлы в сети, которые находятся на большом удалении от других её элементов.

## Диаметр графа

>**Диаметр** — это максимальное расстояние между всеми парами вершин в графе.

Расстоянием между двумя вершинами является длина кратчайшего пути между этими вершинами. Другими словами, чтобы найти диаметр графа, нужно определить самый длинный из всех кратчайших путей между парами вершин в графе. Мы также можем сказать, что диаметр графа — это максимальный эксцентриситет в графе.

````python
import networkx as nx
import pandas as pd

K = nx.Graph()
K.add_edges_from([
    ('A', 'B'), 
    ('B', 'C'), 
    ('B', 'D'), 
    ('C', 'E'), 
    ('C', 'F'), 
    ('D', 'E'), 
    ('E', 'F')
])

adjacency_matrix = nx.adjacency_matrix(K)

# Превращаем в плотный массив NumPy (чтобы можно было сделать DataFrame)
adjacency_array = adjacency_matrix.toarray()

# Делаем DataFrame: строки и столбцы — это узлы графа в порядке K.nodes()
adjacency_df = pd.DataFrame(
    adjacency_array,
    index=list(K.nodes()),
    columns=list(K.nodes())
)

adjacency_df
````

## Матрица расстояний

Когда мы работаем с сетями вершин и рёбер, часто возникает задача определения кратчайших путей между различными точками сети. Для удобства представления и анализа такой информации используется матрица расстояний.

>**Матрица расстояний** — это таблица, показывающая кратчайшее расстояние между каждой парой вершин в графе. Каждая строка и столбец в этой таблице соответствуют вершине в графе, а на пересечении строки и столбца находится значение, показывающее кратчайшее расстояние между этими двумя вершинами. Это позволяет легко оценить и сравнить расстояния внутри сети, облегчая тем самым планирование и анализ.

Помимо транспортных задач эти знания можно применить практически повсеместно. В телекоммуникационных и компьютерных сетях эти метрики могут помочь оптимизировать передачу данных, распределение ресурсов и обеспечение устойчивости сети.

Например, вершины с минимальным эксцентриситетом могут служить эффективными узлами для централизации данных или услуг. Анализ социальных сетей использует эти метрики для идентификации ключевых участников, например, наиболее влиятельных лиц, основываясь на их близости к другим участникам, или для изучения степени связности сообществ. В биологических и экологических сетях, например, в сетях взаимодействия видов, эксцентриситет может указывать на уязвимость видов, а диаметр и радиус — на устойчивость экосистемы.

Кроме того, понимание эксцентриситета, радиуса и диаметра графа может быть полезно при использовании алгоритмов обхода графа и алгоритмов поиска кратчайшего пути, так как эти метрики предоставляют ценную информацию о структуре графа.

# <center> 4. BFS. Поиск в ширину

В предыдущих юнитах курса мы познакомили вас с базовыми понятиями графов, их свойствами и методами анализа. Теперь настало время перейти к практическим алгоритмам, которые применяются для навигации и поиска в этих сложных структурах. Среди них — алгоритм «Поиск в ширину» (BFS). Это важный инструмент любого специалиста, работающего с данными и графами.

>**Алгоритм поиска в ширину** — один из основных методов обхода графа. Его ценность заключается в способности находить кратчайший путь в невзвешенном графе и визуализировать процесс пошагово.

Этот алгоритм не просто ищет путь от одной точки к другой; он стратегически исследует граф слой за слоем, гарантируя, что сначала мы изучим все точки, находящиеся на одинаковом расстоянии от начальной вершины.

На первом этапе работы алгоритма мы выбираем начальную точку и помещаем её в пустую очередь. Затем мы помечаем первую вершину, чтобы избежать её повторного посещения.

Далее, пока очередь не пуста, извлекаем вершину из начала очереди (слева) и просматриваем каждого соседа вершины. Если соседняя вершина не была посещена, помечаем её как посещённую и добавляем в конец очереди.

Алгоритм продолжается до тех пор, пока очередь не опустеет. Опустошение очереди означает, что все вершины, достижимые из начальной точки, были посещены.

Сначала посещаются все вершины на расстоянии 1 от начальной точки, затем все вершины на расстоянии 2 и так далее. Это гарантирует нахождение кратчайших путей в невзвешенном графе, то есть по количеству рёбер.

## Реализация алгоритма поиска в ширину

## Реализация алгоритма поиска в ширину с помощью Python

Давайте подробно рассмотрим реализацию алгоритма поиска в ширину для невзвешенного графа, используя Python и словарь множеств в качестве структуры данных для представления графа.

В невзвешенном графе рёбра равнозначны и не имеют связанных с ними весов. Граф может быть представлен в виде списка смежности, где для каждой вершины указан список вершин, с которыми она непосредственно соединена.

В Python граф обычно представляется в виде словаря, где ключами являются вершины, а значениями — списки смежных вершин.

Для начала определим граф в виде словаря множеств, где ключами будут вершины, а значениями — множества вершин, с которыми они прямо соединены:

````python
adj_list = {
    'A': ['C', 'B', 'K'], # A связана с C, B, K
    'C': ['A', 'B', 'D'], # C связана с A, B, D
    'B': ['A', 'C', 'K'], # B связана с A, C, K
    'K': ['A', 'B', 'F'], # K связана с A, B, F
    'F': ['K', 'D'],      # F связана с K, D
    'D': ['C', 'F']       # D связана с C, F
}
````

Далее мы инициализируем основные структуры данных, используемые в алгоритме: множество для отслеживания посещённых вершин, очередь для хранения вершин, которые предстоит посетить, а также словари для хранения уровней каждой вершины и родителей для возможности восстановления пути.

````python
from collections import deque

def bfs(graph, start_vertex):
    visited = set()  # Посещённые вершины
    queue = deque([start_vertex])  # Очередь для BFS
    levels = {start_vertex: 0}  # Уровни вершин относительно стартовой
    parents = {start_vertex: None}  # Родители вершин в дереве BFS
````

Затем мы начинаем обходить граф, используя BFS. Цель обхода — посетить все вершины графа, доступные из начальной точки, сохраняя при этом информацию, как далеко каждая вершина находится от начальной, и как мы дошли до каждой вершины.

Мы начинаем с извлечения вершины из начала очереди. Эта вершина становится текущей вершиной, которую мы обрабатываем. В начале алгоритма это будет стартовая вершина.

Для текущей вершины мы перебираем всех её соседей. Соседи определяются на основе списка смежности: это все вершины, непосредственно соединённые с текущей.

Если соседняя вершина ещё не была посещена:
* Мы отмечаем её как посещенную, добавляя в множество visited. Это предотвращает повторную обработку вершины в будущем.
* Добавляем соседа в конец очереди. Это гарантирует, что все соседи текущей вершины будут обработаны, прежде чем мы перейдём к вершинам следующего уровня.
* Устанавливаем уровень соседа на один выше, чем уровень текущей вершины. Это отражает минимальное количество шагов (рёбер), необходимых для достижения соседа из начальной точки.
* Записываем текущую вершину как родителя соседа в словарь parents. Это позволяет восстановить путь от начальной вершины к любой другой, следуя по родительским ссылкам.

````python
from collections import deque

adj_list = {
    'A': ['C', 'B', 'K'], # A связана с C, B, K
    'C': ['A', 'B', 'D'], # C связана с A, B, D
    'B': ['A', 'C', 'K'], # B связана с A, C, K
    'K': ['A', 'B', 'F'], # K связана с A, B, F
    'F': ['K', 'D'],      # F связана с K, D
    'D': ['C', 'F']       # D связана с C, F
}

def bfs(graph, start_vertex):
    visited = set()  # Посещённые вершины
    queue = deque([start_vertex])  # Очередь для BFS
    levels = {start_vertex: 0}  # Уровни вершин относительно стартовой
    parents = {start_vertex: None}  # Родители вершин в дереве BFS

    while queue:
        vertex = queue.popleft()
        visited.add(vertex)

        for neighbor in graph[vertex]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
                levels[neighbor] = levels[vertex] + 1
                parents[neighbor] = vertex

    return levels, parents

levels, parents = bfs(adj_list, 'A')
````

````python
print("Порядок обхода:", list(bfs(adj_list, 'A')[0]))
print("Уровни:", levels)
print("Родители:", parents)
````

Математический подход может быть неэффективным или даже невозможным для применения к большим графам вручную. Компьютерные алгоритмы могут обрабатывать графы с тысячами и миллионами вершин и рёбер. Поэтому важно понимать оба подхода: первый для понимания сути алгоритма, а второй — для быстрой работы с большими массивами данных.

Используя алгоритм BFS, мы разработали план, по которому сообщение будет распространяться от сервера к серверу, чтобы минимизировать общее время рассылки. При этом мы выстроили оптимальный порядок оповещения серверов таким образом, чтобы каждый последующий сервер получал сообщение как можно быстрее.

Вы узнали, как инициализировать процесс поиска, создавая очередь для последовательного изучения вершин и используя множество для отслеживания уже посещённых точек.

Запомните, что сложные задачи маршрутизации могут быть разложены на простые действия: последовательное добавление вершин в очередь и их исключение после обработки превращается в логическую цепочку, по которой можно последовать от начала до конца.

BFS может использоваться, например, для определения минимального числа остановок в системе общественного транспорта, когда важно учитывать только количество пересадок, а не длительность поездки.

Важно помнить и о недостатках BFS: он не подходит для взвешенных графов, где рёбра имеют разную стоимость или длину. В таких случаях другие алгоритмы, как алгоритм Дейкстры, могут предоставить более точные результаты для нахождения кратчайшего пути.

# <center>  5. DFS. Поиск в глубину

В рамках этого юнита мы рассмотрим другой алгоритм: он поможет нам в определении связности, топологической сортировке и нахождении компонент сильной связности. Такой алгоритм получил название «Поиск в глубину» (DFS, Depth-First Search).

>**DFS** — это алгоритм, который обходит граф таким образом, что он пытается идти как можно дальше по каждой ветви перед тем, как отступить. Эта характеристика делает его идеальным для задач, где нужно исследовать все возможные пути или конфигурации.

Основная идея рекурсивного DFS заключается в использовании стека вызовов функций для отслеживания пути обхода. Начиная с выбранной вершины, алгоритм рекурсивно исследует каждый путь до тех пор, пока не достигнет конца или не обойдёт все возможные пути от этой вершины. Вот как это выглядит:
1. Выбираем вершину для начала обхода и вызываем для неё рекурсивную функцию DFS.
2. Помечаем текущую вершину как посещённую, чтобы не вернуться к ней снова.
3. Для каждого соседа выбранной вершины, который ещё не был посещён, рекурсивно вызываем DFS.
4. Когда достигнут конец пути или все соседи посещены, происходит откат к предыдущей вершине — рекурсия возвращается на уровень выше.
5. Рекурсия завершается, когда все вершины, доступные из начальной точки, будут исследованы.

Давайте рассмотрим алгоритм на том же примере, что и в прошлый раз.

## Реализация алгоритма поиска в глубину с помощью Python

Итак, нам надо идти вглубь графа настолько, насколько это возможно: на каждом шаге будем переходить к следующей вершине до тех пор, пока не будет достигнут конец пути или не будут посещены все вершины. После чего алгоритм отступает и исследует следующие пути.

На первом шаге мы также определяем граф, с которым мы будем работать, в виде списка смежности:

Реализуем функцию dfs(), которая обеспечит рекурсивный обход графа из заданной начальной вершины. Перед началом обхода необходимо инициализировать структуры данных, которые будут использоваться в процессе работы алгоритма:

* visited — множество для отслеживания посещенных вершин.
* parents — словарь, где ключи — вершины, а значения — родительские вершины, из которых в данную вершину можно попасть. Для начальной вершины родителем является None.
* path — список, в который будут добавляться вершины в порядке их посещения.

```python

Сначала проверяем, была ли текущая вершина node уже посещена. Если нет, то продолжаем обработку.
```python

```

Добавляем текущую вершину в множество посещённых, чтобы избежать её повторного обхода в будущем.
```python

```
Вносим текущую вершину в список path, фиксируя тем самым порядок обхода.
```python

```
Итерируем по всем соседям текущей вершины. Соседи определяются списком, ассоциированным с ключом node в словаре graph.
```python

```
Для каждого соседа, который еще не был посещён, запоминаем его родителя (текущую вершину) и рекурсивно вызываем dfs(), продвигаясь глубже в граф.
```python
if neighbor not in visited:
    parents[neighbor] = node
    dfs(visited, graph, neighbor, path, parents)
```

Тогда полный код будет выглядеть следующим образом:

````python
def dfs(visited, graph, node, path, parents):
    if node not in visited:
        visited.add(node)  # Помечаем вершину как посещённую
        path.append(node)  # Добавляем вершину в текущий путь
        for neighbor in graph[node]:
            if neighbor not in visited:
                parents[neighbor] = node  # Запоминаем родителя для соседа
                dfs(visited, graph, neighbor, path, parents)

adj_list = {
    'A': ['B', 'C', 'K'],
    'B': ['A', 'C', 'K'],
    'C': ['A', 'B', 'D', 'F'],
    'D': ['C', 'F'],
    'F': ['K', 'D'],
    'K': ['A', 'B', 'F']
}

visited = set()
parents = {'A': None}  # Начальная вершина не имеет родителя
path = []

dfs(visited, adj_list, 'A', path, parents)

print("Итоговый путь:", path)
print("Родители:", parents)
````

Особенность DFS — в его рекурсивной природе и использовании стека вызовов для отслеживания пути обхода. Это делает его ценным для ряда задач на графах, таких как определение связности, топологическая сортировка, поиск компонент сильной связности и многих других.

Результат работы алгоритма показал, что при правильном подходе и понимании логики графов можно эффективно решать сложные задачи, связанные со структурами данных. Итоговый путь, полученный в результате обхода, и структура родителей для каждой вершины, позволяют не только понять структуру графа, но и найти оптимальные пути, раскрыть связи между элементами и даже восстановить полный маршрут обхода.

# <center> 6. Алгоритм Дейкстры

В предыдущих юнитах мы уже рассматривали алгоритм, позволяющий найти путь с наименьшим количеством рёбер — поиск в ширину. Но что, если нам нужно найти кратчайший путь для взвешенного графа? Не факт, что даже при наименьшем числе рёбер путь будет самым коротким. Тут на помощь приходит алгоритм Дейкстры.

Алгоритм, о котором будет идти речь в этом юните, был предложен Эдсгером Дейкстрой в 1959 году и с тех пор находит применение во многих областях, например, компьютерных сетях и при планировании маршрутов.

Алгоритм Дейкстры ищет кратчайший путь от начальной вершины до всех остальных в графе. Идея состоит в последовательном рассмотрении всех возможных путей до каждой вершины и выборе самого короткого.

Алгоритм использует принцип жадности: на каждом шаге выбирается вершина, наиболее близкая к начальной, и рассматриваются все её соседи.

>**Жадные алгоритмы** — это класс алгоритмов для решения вычислительных задач, основанный на принципе жадности. Принцип жадности заключается в выборе локально оптимальных решений на каждом шаге с надеждой, что эти локальные оптимумы приведут к глобально оптимальному решению всей задачи.

Запомните следующие особенности жадных алгоритмов:
* **Локально оптимальный выбор**

На каждом этапе алгоритма делается выбор, который кажется лучшим в данный момент, без учёта будущих последствий этого выбора.
* **Невозвратность**

Жадные алгоритмы не рассматривают возвращение к предыдущим шагам для пересмотра решений, даже если это может привести к улучшению общего результата.
* **Простота и эффективность**

Благодаря своей прямолинейности, жадные алгоритмы часто оказываются более простыми в реализации и быстрее в исполнении, чем другие алгоритмические подходы, например, динамическое программирование или методы полного перебора.

Представим, что мы — транспортная компания, и наша основная задача — эффективная доставка грузов. На этот раз мы сталкиваемся с задачей оптимизации маршрута от города Владимир до Ростова Великого.

Города представляют собой узлы, связанные множеством дорог, которые наша компания может использовать для перевозки товаров. Для упрощения планирования маршрутов и уменьшения транспортных издержек мы хотим применить алгоритм Дейкстры, чтобы найти кратчайший путь между этими двумя городами.

Алгоритм Дейкстры идеально подходит для решения таких задач, так как он позволяет вычислить кратчайшие пути от одного узла — города — до всех других в графе с положительными весами рёбер — расстояниями между городами.

Вначале алгоритм устанавливает начальные значения расстояний от исходного города до всех остальных как бесконечности; расстояние до исходного узла приравнивает нулю.

## Процесс работы алгоритма Дейкстры

Опишем, как работает этот алгоритм (рисунок 19).
1. Устанавливаем расстояние от Владимира до самого себя равным нулю, а до всех других городов — бесконечно.
2. Последовательно посещаем все города, начиная с тех, что имеют наименьшее текущее известное расстояние от Владимира.
3. Каждый раз, когда мы посещаем город, мы обновляем расстояние от Владимира до всех его соседей. Если расстояние через посещённый город короче, чем уже известное расстояние, мы обновляем значение.
4. Продолжаем этот процесс до тех пор, пока все города не будут посещены и наименьшее расстояние до каждого из них не будет найдено.

Сфер применения алгоритма Дейкстры много: это и геоинформационные системы, и финансовый анализ, и распределённые вычисления.

Запомните, что в отличие от алгоритма поиска в ширину, который идеально подходит для невзвешенных графов и находит кратчайший путь по количеству рёбер, алгоритм Дейкстры учитывает именно суммарный вес пути. Он исследует все вершины графа, постепенно увеличивая количество изученных вершин и выбирая для каждой из них наименьшее известное расстояние от начальной точки.

Алгоритм Дейкстры может быть реализован с использованием различных структур данных: массивы, связные списки и бинарные кучи.

Заметим, что алгоритм Дейкстры эффективно работает только в графах без отрицательных весов рёбер, поскольку наличие отрицательных весов может привести к появлению циклов отрицательной стоимости, в которых алгоритм не сможет корректно функционировать.
