# <center> СДНФ (совершенная дизъюнктивная нормальная форма)

```python
c = 1        # наняли в первый год
s = 1        # всего сотрудников

print(f"Год 1: набрали {c}, всего {s}")

for y in range(2, 21):
    c *= 2   # каждый год вдвое больше, чем в предыдущий
    s += c   # добавляем к общему счётчику
    print(f"Год {y}: набрали {c}, всего {s}")

print(f"\nИтого за 20 лет нанято: {s}")
```

```python
s = 2**20 - 1   # b1*(q^n - 1)/(q - 1), где b1=1, q=2, n=20
print(s)
```
