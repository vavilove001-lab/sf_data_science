>Проанализируйте датасет на наличие дублирующихся винных обзоров. Если дублирующиеся записи есть, удалите их. В поле ниже введите, сколько дублирующихся записей вам удалось обнаружить.

````python
dupl_columns = list(data.columns)

mask = data.duplicated(subset=dupl_columns)
duplicates = data[mask]
print(f'Число найденных дубликатов: {duplicates.shape[0]}')
````

````python
data_dedupped = data.drop_duplicates(subset=dupl_columns)
print(f'Результирующее число записей: {data_dedupped.shape[0]}')
````

>Проверьте датасет на наличие пропусков в данных. В каких из представленных признаках были обнаружены пропуски?

````python
cols_null_percent = data_dedupped.isnull().mean() * 100
cols_with_null = cols_null_percent[cols_null_percent>0].sort_values(ascending=False)
display(cols_with_null)
````

````python
#задаем минимальный порог: вычисляем 70% от числа строк
thresh = data_dedupped.shape[0]*0.7
#удаляем столбцы, в которых более 30% (100-70) пропусков
data_dedupped = data_dedupped.dropna(thresh=thresh, axis=1)
````

````python
cols_null_percent = data_dedupped.isnull().mean() * 100
cols_with_null = cols_null_percent[cols_null_percent>0].sort_values(ascending=False)
display(cols_with_null)
````

````python
#создаем копию исходной таблицы
fill_data = data_dedupped.copy()
#создаем словарь имя столбца: число(признак) на который надо заменить пропуски
values = {
    'designation': 'unknown',
    'taster_twitter_handle': 'unknown',
    'taster_name': 'unknown',
    'region_1': 'unknown',
    'price': fill_data['price'].mean(),
    'country': 'US',
    'province': 'California',
    'variety': 'Pinot Noir'
}
#заполняем пропуски в соответствии с заявленным словарем
fill_data = fill_data.fillna(values)
#выводим результирующую долю пропусков
fill_data.isnull().mean()
````

````python
# Сохраняем таблицу для дальнейшей работы
fill_data.to_csv('data/cleaned_data.csv', index=False, sep=';')
````
