## <center> Подготовка данных к обучению модели

## Этап 1. Первичный анализ данных
#### Для всех типов признаков:
* импорт данных (pd.read_csv(), pd.read_excel() и т. д.);
* оценка структуры: df.info(), df.shape;
* просмотр первых строк: df.head();
* проверка распределения: df.describe() для числовых, value_counts() для категориальных.

## Этап 2. Очистка данных
#### Общие задачи:
* удаление дубликатов: df.drop_duplicates();
* выявление и обработка выбросов (для числовых);
* обработка пропусков (для обоих типов).

#### Для числовых признаков:
* заполнение пропусков: средним, медианой или константой;
* обнаружение выбросов через IQR‑метод, Z‑score или визуализацию (boxplot);
* удаление или корректировка выбросов.

#### Для категориальных признаков:
* заполнение пропусков: модой (mode()), специальной категорией ('Unknown') или на основе других признаков;
* унификация категорий: приведение к единому регистру, исправление опечаток;
* объединение редких категорий (если доля < 1–5 %).

## Этап 3. Преобразование категориальных признаков
### Методы кодирования:
#### 1. Label Encoding — присвоение числовых меток:
* подходит для порядковых категорий ('низкий' → 0, 'средний' → 1, 'высокий' → 2);
* реализация: sklearn.preprocessing.LabelEncoder или pd.factorize().

#### 2. One‑Hot Encoding — создание бинарных столбцов:
* для номинальных категорий без порядка;
* реализация: pd.get_dummies() или sklearn.preprocessing.OneHotEncoder;
* риск: рост размерности при большом числе категорий.

#### 3. Target/Mean Encoding — замена на среднее значение целевой переменной по категории:
* требует осторожности (риск переобучения);
* применимо в задачах классификации/регрессии.

#### 4. Binary Encoding — кодирование в двоичном представлении:
* компромисс между Label и One‑Hot;
* меньше размерность, чем у One‑Hot.

## Этап 4. Обработка числовых признаков
#### Преобразования:
* **нормализация (MinMaxScaler)** — приведение к диапазону [0, 1];
* **стандартизация (StandardScaler)** — центрирование вокруг 0 с единичным отклонением;
* **дискретизация** — разбиение на интервалы (например, возраст → 'молодой', 'средний', 'пожилой');
* **трансформация распределения** — логарифмирование, Box‑Cox для снижения асимметрии.

## Этап 5. Создание новых признаков (Feature Engineering)
#### Идеи для обоих типов:
* комбинации признаков (например, price / area → price_per_sqm);
* извлечение информации из дат (день недели, месяц, сезон);
* агрегированные статистики по группам (среднее, медиана, количество);
* полиномиальные признаки (квадраты, произведения) для линейных моделей.

## Этап 6. Отбор признаков
#### Методы:
* статистические тесты (хи‑квадрат для категориальных, ANOVA для числовых);
* корреляционный анализ (исключить сильно коррелирующие признаки);
* важность признаков из моделей (например, RandomForest);
* методы уменьшения размерности (PCA, t‑SNE).

## Этап 7. Разделение данных
* разделение на обучающую и тестовую выборки (train_test_split);
* при необходимости — кросс‑валидация (KFold, StratifiedKFold);
* важно: все преобразования (масштабирование, кодирование) применять отдельно к train и test, чтобы избежать утечки данных.

## Этап 8. Сборка пайплайна (Pipeline)
Объединение шагов в единый конвейер:
* гарантирует воспроизводимость;
* предотвращает утечки данных;
* упрощает применение к новым данным.
