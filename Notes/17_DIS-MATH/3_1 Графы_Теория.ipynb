# <center> 3. Виды графов. Неориентированный граф

Для начала разделим графы на две большие категории: **неориентированные** и **ориентированные**.

>В первой категории рёбра графов не имеют направления. Примером могут послужить дорожные сети, где дороги позволяют двигаться в обе стороны.

>**Ориентированный граф** или **орграф** содержит рёбра с определённым направлением, представленные упорядоченными парами вершин. Они идеально подходят для представления асимметричных отношений, где направленность имеет значение. Примером служат страницы в интернете, связанные гиперссылками: они направлены от одной страницы к другой.

## Мультиграф

Ранее мы рассказывали о **кратных рёбрах** — это когда два или более ребра соединяют одну и ту же пару вершин.

>Напомним, что в простом графе каждая пара вершин может быть соединена максимум одним ребром, а петли не разрешены. Однако, когда в структуре допускаются кратные рёбра — то есть, между двумя вершинами может быть более одного ребра — такой граф называется **мультиграфом**.

Кратные рёбра вносят дополнительную сложность в анализ графа, поскольку они создают избыточные связи между вершинами, которые могут затруднить понимание структуры графа и влиять на работу различных алгоритмов, например, при вычислении кратчайших путей.

## Псевдограф и подграф

>**Псевдограф** - это разновидность графа, в котором разрешено наличие как петель, так и кратных рёбер.

>Граф же с петлями и кратными ребрами как на рисунке 7 будет называться **псевдомультиграфом**.

## Гиперграф и гиперребро

Мы уже рассмотрели основные концепции графов, такие как вершины и рёбра, посмотрели, что у нас могут быть кратные ребра и петли. Однако, для моделирования более сложных взаимосвязей между элементами, требуются более продвинутые структуры. В этом контексте особенно выделяются гиперграфы и гиперрёбра.

>**Гиперграф** — это расширенная структура графа, в котором рёбра могут соединять любое количество вершин, а не только две, как в обычных графах. Элементы гиперграфа состоят из вершин и гиперрёбер.

>**Гиперребро** — это набор вершин, и он может соединять две и более вершины, в отличие от рёбер в простых графах, которые соединяют ровно две вершины.

## Взвешенный граф

Может сложиться такая ситуация, что нам, кроме самого факта наличия связи, нужно изучить ещё что-то. Например, мы хотим рассчитать самый эффективный путь доставки товаров или любую другую метрику, которая количественно характеризует связь между вершинами. Тогда на помощь нам придут взвешенные графы.

>**Взвешенный граф** — это особый тип графа, в котором каждое ребро имеет присвоенное ему числовое значение или вес. Эти веса могут представлять различные характеристики или параметры, такие как длина, стоимость, время или ёмкость, в зависимости от контекста задачи или приложения.

## Регулярный граф

После того как мы рассмотрели планарные графы, гиперграфы, взвешенные графы и их особенности, обратим ваше внимание на ещё один важный тип графов — регулярные графы. Эти графы играют ключевую роль в теории графов благодаря их однородной структуре, которая приводит к интересным математическим свойствам и применениям.

>**Регулярный граф** — это тип графа, в котором каждая вершина имеет одинаковую степень, то есть одинаковое количество рёбер, связанных с каждой вершиной. Степень вершин в этом случае постоянна и одинакова для всех вершин графа.

Например, самыми простыми примерами регулярного графа являются треугольник и куб. Другим известным примером 3-регулярного графа является Граф Петерсена. Он состоит из 10 вершин и 15 рёбер. В Графе Петерсена каждая вершина соединена ровно с тремя другими вершинами.



## Двудольный граф

Часто нам нужно моделировать ситуации, где элементы можно разделить на два непересекающихся набора таким образом, что каждое ребро соединяет элементы из разных наборов. В этом нам поможет двудольный граф.

>**Двудольный граф** — это такой граф, множество вершин которого можно разделить на два непересекающихся подмножества так, что каждое ребро соединяет вершины из разных подмножеств.

# <center> 4. Виды графов. Ориентированный граф

Существуют ситуации, для работы с которыми неориентированные графы не подходят. Тогда на помощь приходят графы, в которых заранее задано направление — орграфы. Приведём несколько примеров их использования:
* В информационных технологиях орграфы используются для моделирования направленных потоков данных, например, архитектуры программных систем, где данные перемещаются от одного компонента к другому, следуя определённым правилам и ограничениям. Это позволяет разработчикам визуализировать и анализировать обработку данных и взаимодействие между частями системы.
* Орграфы могут представлять собой сети дорог или маршрутов, где дороги имеют направление движения. Это позволяет анализировать и оптимизировать транспортные потоки, например, для уменьшения пробок или оптимизации маршрутов доставки.
* В социальных науках орграфы могут использоваться для изображения асимметричных социальных отношений, таких как иерархии влияния или паттерны коммуникации, где одно направление связи является доминирующим. Это помогает исследователям визуализировать и анализировать структуру социальных сетей и отношений.

>**Смежность в вершине в орграфе** — это общее количество инцидентных ей дуг, и она рассчитывается как сумма полустепени исхода и полустепени захода этой вершины.

Когда мы анализируем орграфы, такие как транспортные сети, социальные сети или другие системы, где направление имеет значение, эти понятия становятся основой для многих алгоритмов и аналитических методов.

Например, в транспортной сети смежность может указывать на загруженность станции в зависимости от количества прибывающих и отправляющихся маршрутов.

В социальных сетях высокая смежность узла может указывать на значимость человека или организации в коммуникационной сети.

Орграфы могут быть использованы в самых разных областях: от информационных технологий для моделирования потоков данных между компонентами систем, в транспортных сетях для анализа и оптимизации потоков движения, до социальных наук для изучения асимметричных отношений и структур влияния в социальных сетях.
