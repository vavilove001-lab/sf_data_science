Создайте с помощью функции-преобразования новый признак **"Образование"**, который должен иметь 4 категории: "высшее", "неоконченное высшее", "среднее специальное" и "среднее".

Выполните преобразование, ответьте на контрольные вопросы и удалите признак "Образование и ВУЗ".

Совет: обратите внимание на структуру текста в столбце **"Образование и ВУЗ"**. Гарантируется, что текущий уровень образования соискателя всегда находится в первых 2ух слов и начинается с заглавной буквы. Воспользуйтесь этим.

*Совет: проверяйте полученные категории, например, с помощью метода unique()*


```python
def education(option):    
# На вход получает строку из таблицы из столбца "Образование и Вуз"
# Изменяем регистр строки на нижний
# Преобразуем список в строку с разделителем " "
# Получаем срез первые 2 элемента списка и преобразуем обратно в строку
    option = option.lower()
    option_list = option.split(' ')
    option_list = option_list[:2]
    education_type = ' '.join(option_list)
# Проверяем условие на те типы, где нужно вывести 1 слово
# В противном случае возвращаем готовый результат
    if education_type == 'среднее образование':
        return 'среднее'
    elif education_type == 'высшее образование':
        return 'высшее'
    else:
        return education_type
```

* Признак пола должен иметь 2 уникальных строковых значения: 'М' - мужчина, 'Ж' - женщина.

```python
# Функция для определения пола
def gender(option):    
# На вход получает строку из таблицы из столбца "Пол, возраст"
# Преобразуем список в строку с разделителем ","
# Получаем срез первый 1 элемента списка
    option_list = option.split(',')
    gender_var = option_list[0]
    return gender_var[0]
```

```python
# Функция для получения возраста
def age(option):    
# На вход получает строку из таблицы из столбца "Пол, возраст"
# Преобразуем список в строку с разделителем ","
# Получаем  2 элемент списка- возраст
# Из этого элемента списка вычленяем число
    option_list = option.split(',')
    age = option_list[1]
    age = ''.join(filter(str.isdigit, age))

    return int(age)
```

3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…
    * Учитывайте эту особенность в вашем коде

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
def work_experience(str):
    if str is np.nan or str == 'Не указано':
        return None
    
    year_in_words=['год', 'года', 'лет']
    month_in_words=['месяц', 'месяца', 'месяцев']
    
    str_splitted = str.split(' ')[:7]
    years = 0
    months = 0
    for index, item in enumerate (str_splitted):
        if item in year_in_words:
            years = int(str_splitted[index-1])
        if item in month_in_words:
            months = int(str_splitted[index-1])
            
    return int(years*12 + months)
```

```python
# Применяем функцию к столбцу "Опыт работы" и создаем новый признак "Опыт работы (месяц)"
hh_database['Опыт работы (месяц)'] = hh_database['Опыт работы'].apply(work_experience)
```

```python
print('Медианный опыт работы (в месяцах) в таблице равен', round(hh_database['Опыт работы (месяц)'].median()))
```

```python
# Удаляем столбец
hh_database.drop('Опыт работы', axis=1, inplace=True)
```

*Совет: обратите внимание на то, что структура текста может меняться в зависимости от указания ближайшего метро. Учите это, если будете использовать порядок слов в своей программе.*


```python
def city(str):
    million_cities = ['Новосибирск', 'Екатеринбург','Нижний Новгород','Казань', 'Челябинск','Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж','Волгоград']
    cities = str.split(' , ')[0]
    if cities == 'Москва':
        return 'Москва'
    elif cities == 'Санкт-Петербург':
        return 'Санкт-Петербург'
    elif cities in million_cities:
        return 'город-миллионник'
    else:
        return 'другие'
```

```python
hh_database['Город'] = hh_database['Город, переезд, командировки'].apply(city)
```

```python
def relocation(str):
    options = str.split(' , ')[1]
    options_splitted = options.split(' ')[0]
# делаем проверку, если 2 элемент списка - метро, переходим к 3 элементу списка и работаем с ним
    if options_splitted == 'м.':
        new_options = str.split(' , ')[2]
        new_options_splitted = new_options.split(' ')[0]
        if new_options_splitted == 'не':
            return False
        else:
            return True
# во всех остальных случаях работаем со 2 элементом списка
    else:
        if options_splitted == 'не':
            return False
        else:
            return True
```

```python
# Проверяем работу функции
relocation('Москва , м. Беломорская , готов к переезду, не готов к командировкам')
```

```python
hh_database['Готовность к переезду'] = hh_database['Город, переезд, командировки'].apply(relocation)
```

```python
def business_trip(str):
# делаем проверку, если 2 элемент списка - метро, переходим к 4 элементу списка и работаем с ним
    str_list = str.split(' , ')
    # Убедимся, что список достаточно длинный, чтобы избежать IndexError, 
    # для случая когда данные о коммандировках потерялись
    if len(str_list) < 3: 
        return False
    # Делаем проверку, есть ли метро в списке, чтобы понимать к какому элементу списка обращаться
    if len(str_list) == 4:
        trip_options = str.split(' , ')[3]
        trp_options_splitted = trip_options.split(' ')[0]
        if trp_options_splitted == 'не':
            return False
        else:
            return True
# во всех остальных случаях работаем с 3 элементом списка
    else:
        t_options = str.split(' , ')[2]
        t_options_splitted = t_options.split(' ')[0]
        if t_options_splitted == 'не':
            return False
        else:
            return True
```

```python
# Проверяем работу функции
business_trip('Москва ,  м. Площадь Ильича , готова к переезду (Москва) , не готова к редким командировкам ')
```

5. Рассмотрим поближе признаки **"Занятость"** и **"График"**. Сейчас признаки представляют собой набор категорий желаемой занятости (полная занятость, частичная занятость, проектная работа, волонтерство, стажировка) и желаемого графика работы (полный день, сменный график, гибкий график, удаленная работа, вахтовый метод).
На сайте hh.ru соискатель может указывать различные комбинации данных категорий, например:
* полная занятость, частичная занятость
* частичная занятость, проектная работа, волонтерство
* полный день, удаленная работа
* вахтовый метод, гибкий график, удаленная работа, полная занятость

Такой вариант признаков имеет множество различных комбинаций, а значит множество уникальных значений, что мешает анализу. Нужно это исправить!

Давайте создадим признаки-мигалки для каждой категории: если категория присутствует в списке желаемых соискателем, то в столбце на месте строки рассматриваемого соискателя ставится True, иначе - False.

Такой метод преобразования категориальных признаков называется One Hot Encoding и его схема представлена на рисунке ниже:
<img src=https://raw.githubusercontent.com/AndreyRysistov/DatasetsForPandas/main/ohe.jpg>
Выполните данное преобразование для признаков "Занятость" и "График", ответьте на контрольные вопросы, после чего удалите их из таблицы

```python
def employment(row):
    text = row['Занятость']
    # обработка (можно сразу вернуть кортеж)
    return ( 'полная занятость' in text,
             'частичная занятость' in text,
             'проектная работа' in text,
             'стажировка' in text,
             'волонтерство' in text )
```

```python
new_cols = ['полная занятость', 'частичная занятость', 'проектная работа', 'стажировка', 'волонтерство']
hh_database[new_cols] = hh_database[['Занятость']].apply(employment, axis=1, result_type='expand')
```

```python
def schedule(row):
    text = row['График']
    # обработка (можно сразу вернуть кортеж)
    return ( 'полный день' in text,
             'сменный график' in text,
             'гибкий график' in text,
             'удаленная работа' in text,
             'вахтовый метод' in text )
```

```python
new_cols = ['полный день', 'сменный график', 'гибкий график', 'удаленная работа', 'вахтовый метод']
hh_database[new_cols] = hh_database[['График']].apply(schedule, axis=1, result_type='expand')
```

```python
exchange = pd.read_csv('data/ExchangeRates.csv', sep = ',')
```

```python
exchange.drop('per', axis=1, inplace=True)
exchange.drop('time', axis=1, inplace=True)
exchange.drop('vol', axis=1, inplace=True)
exchange.head(3)
```

```python
hh_database['Обновление резюме'] = pd.to_datetime(hh_database['Обновление резюме'], dayfirst=True)
```

```python
hh_database['Обновление резюме'] = hh_database['Обновление резюме'].dt.date
```

```python
exchange['date'] = pd.to_datetime(exchange['date'], dayfirst=True)
```

```python
# Выделяем из столбца ЗП цифру
def sum(str):
    sum = str.split(' ')[0]
    return int(sum)
```

```python
# Заносим в промежуточный столбец цифру зп
hh_database['цифра зп'] = hh_database['ЗП'].apply(sum)
```

```python
def currency(str):
    cur= str.split(' ')[1]
    if cur == 'грн':
        return 'UAH'
    elif cur == 'белруб':
        return 'BYN'
    elif cur == 'сум':
        return 'UZS'
    elif cur == 'руб.':
        return 'RUB'
    else:
        return cur
```

```python
# Заносим в промежуточный столбец валюту, в которой указана зп
hh_database['валюта'] = hh_database['ЗП'].apply(currency)
```
