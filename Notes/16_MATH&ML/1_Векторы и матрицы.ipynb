## ЛИНЕЙНАЯ КОМБИНАЦИЯ ВЕКТОРОВ

>**Линейная комбинация** — это объединение операций сложения векторов и умножения на скаляр в одну. Каждый вектор умножается на соответствующий ему коэффициент, все умноженные векторы складываются.

````python
p = np.array([2, 4, 5])
v = np.array([8, 10, 2])
s = np.array([0, 12, 7])
omega1 = 500
omega2 = 100
omega3 = 0
u = omega1*p + omega2*v + omega3*s
u
````

Мы рассмотрели примеры с прямоугольными матрицами. А что насчёт квадратных матриц? 

С такими матрицами чуть проще. Произведение квадратных матриц существует, когда порядок обеих матриц совпадает. Причём результатом умножения квадратных матриц будет квадратная матрица того же порядка, что и исходные.

Например, матрицу 3х3 можно умножить только на матрицу 3х3, но нельзя на матрицу 4х4 или 4х3 или, тем более, 11х23.
