```python
import pandas as pd
import statistics
import re
```

```python
data = pd.read_csv('data/wine_cleared_2.csv')
```

```python
data.head(3)
```

```python
# Создадим новый признак price_round, означающий округлённую до целого числа цену за бутылку вина:
# для удобства сразу преобразуем признак в int
data['price_round'] = data['price'].round().astype(int)
```

```python
data = data.drop('price', axis = 1)
```

```python
# Выполните код для нахождения года вина при помощи регулярного выражения:
regex = '\d{4}' # регулярное выражение для нахождения чисел
data['year'] = data['title'].str.findall(regex).str.get(0)
```

```python
# Вы можете создать новый бинарный признак is_usa и присвоить ему 1 в случае, если вино произведено в США, иначе — 0.
data['is_usa'] = data['country'].apply(lambda x: 1 if x == 'US' else 0)
```

```python
data['is_france'] = data['country'].apply(lambda x: 1 if x == 'France' else 0)
data['is_italy'] = data['country'].apply(lambda x: 1 if x == 'Italy' else 0)
```

```python
data['old_wine'] = data['year'].apply(lambda x: 1 if float(x) < 2010 else 0)
```

```python
data['old_wine'].sum()
```

```python
import re

def extract_locality(title):
    pattern = r'\(([^)]+)\)$'
    match = re.findall(pattern, title)
    if match:
        return ''.join(match)
    else:
        return None
```

```python
data['locality'] = data['title'].apply(extract_locality)
```
