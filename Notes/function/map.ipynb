### Встроенная функция map()

В Python можно применять функции к элементам списка и без циклов. Для этого используется встроенная функция map(). Использование map() даёт сразу несколько преимуществ:

простота кода — код становится короче, вплоть до одной строки, и выглядит более лаконичным;
скорость — возможности, которые мы рассмотрим, в большинстве случаев работают быстрее, чем использование функции в цикле.

ФУНКЦИЯ КАК ОБЪЕКТ

Важно понять одну вещь: функция, как и всё в Python (кроме ключевых слов, таких как if, def, while и т. д.) — это объект.

Важно! Когда мы обращаемся с функцией как с объектом, мы не пишем скобки после имени функции. Они нужны только тогда, когда мы вызываем функцию и передаем в неё аргументы.

В качестве простейшего примера напишем функцию apply_func(). Она будет иметь два параметра — список x и функция func, которую нужно применить к переданному списку.

```python
# Функция, которая в качестве аргументов принимает 
# Другую функцию func и список x

def apply_func(func, x):
    # Функция вызывает переданную функцию, передавая ей в аргументы список x
    return func(x)
```

Теперь мы можем вызвать функцию apply_func(), передав в качестве её параметра встроенную функцию max() для вычисления максимума и список [1, 10, 35, 20, -1].

```python
print(apply_func(max, [1, 10, 35, 20, -1]))
```

```python
print(apply_func(min, [1, 10, 35, 20, -1]))
```

```python
print(apply_func(len, [1, 10, 35, 20, -1]))
```

 у функции точно так же, как и любого другого объекта, есть свой тип данных — тип function. Давайте убедимся в этом.

### ФУНКЦИЯ MAP

Итак, у нас есть итерируемый объект — пусть это будет список слов — и нам с каждым его элементом необходимо совершить преобразование, например вычислить длину каждого слова в нём, привести все слова к нижнему регистру, удалить знаки препинания и т. д.

Пример №1

Условие задачи

В задачах обработки естественного языка (Natural Language Processing, NLP) иногда имеет значение не только само слово, но и его длина.

Перед нами стоит следующая микрозадача: вычислить длину каждого слова из набора данных, чтобы в дальнейшем использовать эту информацию для анализа теста.

Пусть задан список words_list со словами, длину которых нам надо посчитать.

```python
# Список из слов, длину которых мы хотим вычислить
words_list = ["We're", 'in', 'a', 'small', 'village', 'near', 'Chicago', 'My', "cousin's", 'getting', 'married.']
```

Решение

Составим новый список, в котором будем хранить длины слов, назовём его lengths_list. Самое простое решение — воспользоваться циклом.

```python
# Создаём пустой список, куда будем заносить результаты
lengths_list = []
# Создаём цикл по элементам списка names
for word in words_list:
    # Вычисляем длину текущего слова
    length = len(word)
    # Добавляем вычисленную длину слова в список
    lengths_list.append(length)
# Смотрим, что получилось
print(lengths_list)
```

Приведённый выше способ использует создание нового списка с пополнением его значениями в цикле. Однако есть способ сделать ту же самую операцию буквально одной строкой кода.

Для этого воспользуемся специальной встроенной в Python функцией map(). Она позволяет преобразовать каждый элемент итерируемого объекта по заданной функции.

Аргументы функции map() следующие:

первый — функция, которую необходимо применить к каждому элементу;
второй — итерируемый объект (например, список).
Можно записать это в виде шаблона кода:

### map(< имя_функции>, < итерируемый_объект>)

У нас есть встроенная функция для вычисления длины — len(), она возвращает длину переданного ей объекта. Нужно применить эту функцию к каждому элементу списка words_list. Тогда наш код с применением функции map() будет выглядеть следующим образом:

```python
# Применяем функцию len к каждому элементу списка
# Для этого передаём функцию len и список words_list в функцию map 
lengths_map = map(len, words_list)
# Смотрим, что получилось
print(lengths_map)
```

Как мы видим, функция map() возвращает объект типа данных map.

Примечание. Этот объект на самом деле является разновидностью итератора, но о том, что такое итераторы, как они работают и как создать собственный итератор, мы поговорим в отдельном модуле курса. Сейчас для общего понимания достаточно запомнить, что результаты работы функции map() необходимо будет дополнительно преобразовывать.

Чтобы увидеть результат в виде списка, распакуем содержимое map-объекта в список. Сделаем это с помощью уже знакомой нам встроенной функции list(). Тогда итоговый код будет выглядеть следующим образом:

```python
# Применяем функцию len к каждому элементу списка
# Для этого передаём функцию len и список words_list в функцию map
# Результат преобразуем в список
lengths_list = list(map(len, words_list))
# Посмотрим, что получилось
print(lengths_list)
```

Условие задачи

Представим, что у нас есть список зарплат сотрудников salaries.

Необходимо посчитать, какой налог необходимо удержать с каждой зарплаты. Допустим: у нас используется прогрессивная система налогообложения, то есть чем выше зарплата, тем выше налог. Налог на доход меньше 1000 составляет 5 %, от 1000 до 2000 — 10 %, а если доход превышает 2000, то налог составляет 15 %.

Решение

```python
def calculate_tax(salary):
    if salary < 1000: # Если зарплата < 1000
        # Налог — 5 % от зарплаты
        return salary * 0.05
    elif salary < 2000: # Если зарплата < 2000
        # Налог — 10 % от зарплаты
        return salary * 0.1
    else:
        # Налог — 15 % от зарплаты
        return salary * 0.15
```

```python
print(calculate_tax(1500))
```

Заметьте, что наша функция работает для одного объекта.

Теперь применим эту функцию к каждому элементу списка salaries с помощью функции map():

```python
# Список зарплат
salaries = [1500, 2200, 3500, 1200]
```

```python
# Применяем функцию calculate_tax к каждому элементу списка
# Для этого передаём функцию calculate_tax и список salaries в функцию map() 
# Результат преобразуем к списку
taxes = list(map(calculate_tax, salaries))
# Выведем, что получилось
print(taxes)
```

Пример №4

Условие задачи

Допустим, у нас есть словарь quotes_dict, ключами которого являются названия произведений, а значениями — цитаты из этих произведений.

```python
quotes_dict = {
    'Преступление и наказание': 'А знаешь ли... что низкие потолки и тесные комнаты душу и ум теснят.', 
    'Война и мир': 'И нет величия там, где нет простоты, добра и правды.', 
    'Анна Каренина': 'Если искать совершенства, то никогда не будешь доволен.', 
    'Детство': 'Правил у нас много, а правды нет.', 
    'Колесо времени': 'Ничто так на соединяет людей, как улыбка.'
}
```

В рамках решения задачи анализа естественного языка нам необходимо преобразовать тексты, а именно:

привести к нижнему регистру;
удалить знаки препинания (точки и запятые);
разделить на отдельные слова.

Решение

У нас есть словарь. Нужно написать функцию, которая будет преобразовывать каждый элемент этого словаря, и применить эту функцию.

Однако удобнее работать со словарём, представленным в виде списка кортежей, где первый элемент кортежа — ключ, а второй — значение. Напомним, что для преобразования словаря в список кортежей используется метод item(). Посмотрим на пример:

```python
print(list(quotes_dict.items()))
```

Теперь у нас есть список, состоящий из кортежей: первым элементом кортежа является название произведения, а вторым — цитата.

Для решения задачи через map() необходимо выделить в функцию преобразования для одного итерируемого объекта (у нас это список). В нашем случае элементом списка являются кортежи, поэтому нам нужно написать функцию, которая будет обрабатывать один такой кортеж.

Назовём эту функцию preprocessing_quotes(). Внутри себя функция должна выполнять описанные выше преобразования с кортежем и возвращать новый кортеж, первым элементом которого является исходный ключ, а вторым — преобразованная цитата.

```python
# Функция для преобразования одного элемента списка кортежей
def preprocessing_quoutes(name_quotes_tuple):
    # Разделяем составляющие кортежа на отдельные переменные
    name, text = name_quotes_tuple
    # Приводим цитату к нижнему регистру
    text = text.lower()
    # Заменяем точки и запятые на пустые строки
    text = text.replace('.', '')
    text = text.replace(',', '')
    # Разделяем цитату на слова по пробелу
    words = text.split(' ')
    # Возвращаем кортеж из названия произведения и текста
    return (name, words)
```

```python
# Проверяем, как работает функция
test_text = ('Детство', 'Правил у нас много, а правды нет.')
print(preprocessing_quoutes(test_text))
```

Отлично, всё работает, как мы и задумывали! Осталось лишь применить нашу функцию к каждому элементу списка кортежей. Результат преобразуем сразу в словарь:

```python
# Применяем функцию processing_quoutes к списку кортежей
# Результат преобразовываем в словарь
processed_quotes_dict = dict(map(preprocessing_quoutes, quotes_dict.items()))

print(processed_quotes_dict)
```

## MAP + LAMBDA = ♥

В предыдущем модуле вы узнали о lambda-функциях — особом синтаксисе для записи коротких функций. Напомним, что lambda-функции не ускоряют работу кода, а просто иногда упрощают его написание.

Lambda-функции очень пригодятся нам в этом разделе.

Мы сказали, что можем применять любую функцию в функции map(). А значит, и lambda-функции тоже. Это очень распространённая практика, когда нам необходимо выполнить простейшие действия с элементами последовательностей, которые можно записать в виде одной lambda-функции.

Рассмотрим пример. Пусть у нас есть список из чисел number_list. Необходимо каждое число возвести в куб. Конечно, можно было сделать это в цикле. Можно было бы написать отдельную функцию cube(), которая бы возводила число в куб, и применить эту функцию с помощью map() ко всем элементам списка.

Но зачем плодить лишний код, если задачу можно решить через lambda-функции? Посмотрите:

```python
# Список чисел
number_list = [11, 12, 13, 14, 15, 16]
# Создаём lambda-функцию, которая возводит число в куб, и применяем её к списку
cube_number_list = list(map(lambda x: x**3, number_list))
# Выводим результирующий список
print(cube_number_list)
```

** А вот ещё пример. Есть список строк, которые нужно развернуть (отобразить зеркально). Опять же можно было бы реализовать это через цикл или написать отдельную функцию, однако разворачивание строки — очень простая операция, которую можно вполне поместить в lambda-функцию.

Примечание. Напомним, что развернуть строку можно с помощью среза с отрицательным шагом. Например, если есть строка x, то её зеркальное отображение будет выглядеть вот так: x[::-1].

```python
# Список строк, которые надо развернуть
str_list = ['шалаш', 'казак', 'палиндром', 'рвал дед лавр']
# Создаем lambda-функцию, которая переворачивает одну строку, и применяем её к списку
reverse_str_list = list(map(lambda x: x[::-1], str_list))
# Выводим результирующий список
print(reverse_str_list)
```

** Пример №5

Условие задачи

Есть список кортежей. Один кортеж соответствует одному объекту (в Data Science его часто называется наблюдением) и состоит из трёх элементов (признаков объекта):

имени человека;
его роста (в метрах);
веса (в килограммах).

```python
data = [('Amanda', 1.61, 51), ('Patricia', 1.65, 61), ('Marcos', 1.91, 101)]
```

Необходимо создать список, состоящий из индексов массы тела для каждого человека. Индекс массы тела достаточно будет округлить до одного знака после запятой.

Решение

Итак, начнём с функции, а точнее — с lambda-функции. Она будет принимать на вход кортеж из трёх элементов, а возвращать — новый кортеж из четырёх элементов. В этом кортеже первые три элемента те же, что и в исходном, а четвёртый равен округлённому до одного знака результату деления элемента кортежа с индексом 2 (массы) на квадрат элемента кортежа с индексом 1 (роста):

```python
# Положим lambda-функцию в переменную для удобства
# x — кортеж (элемент исходного списка data)
map_func = lambda x: (x[0], x[1], x[2], round(x[2] / (x[1]**2), 1))
```

```python
# Проверим работу функции для одного кортежа
map_func(('Amanda', 1.61, 51))
```

#### Как вы могли заметить, если первые три элемента нового кортежа остаются неизменными, то прописывать обращение к ним в функции необязательно — мы можем просто распаковать содержимое кортежа. Напомним, что для этого нужно лишь поставить * перед переменной. Тогда код lambda-функции будет выглядеть следующим образом:

```python
# Положим lambda-функцию в переменную для удобства
# x — кортеж (элемент исходного списка data)
map_func = lambda x: (*x, round(x[2] / (x[1]**2), 1))
```

```python
# Проверим работу функции для одного кортежа
map_func(('Amanda', 1.61, 51))
```

Отлично! Осталось только применить функцию ко всем кортежам из списка data и преобразовать результат в список:

```python
# Создаём lambda-функцию, которая считает BMI, и применяем её к каждому элементу списка
map_func = lambda x: (*x, round(x[2]/(x[1]**2), 1))
updated_data = list(map(map_func, data))
# Выводим результат
print(updated_data)
```

Примечание. Выделять отдельную переменную под lambda-функцию необязательно — мы сделали это, чтобы уменьшить поток из открывающихся и закрывающихся скобок и повысить читаемость кода.

Как видите, генерировать новые признаки с помощью map() очень просто. Достаточно лишь написать соответствующую функцию для обработки одного наблюдения и применить её ко всем сразу.

Примечание. Стоит отметить, что работа с реальными данными, как правило, организуется не с помощью списков, а с помощью специальных объектов из библиотеки Pandas, о которой мы поговорим далее. Однако в Pandas есть метод map(), который работает в точности так же, как и встроенная функция map(). Поэтому, если мы разберёмся с особенностями map() сейчас, нам будет проще понять её библиотечный аналог в дальнейшем.

** Допустим, мы решаем задачу оценки стоимости недвижимости.

В списке data представлены усреднённые данные по домам в районах Бостона. Каждый вложенный в список кортеж описывает средние данные по одному району (для примера представлены данные о семи участках). В этом кортеже представлены следующие признаки (в порядке следования):

x₁ — уровень преступности на душу населения по городам;

x₂ — среднее количество комнат в доме;

x₃ — доля зданий, построенных до 1940 г. и занимаемых владельцами;

x₄ — полная ставка налога на имущество за каждые 10 000 долларов стоимости;

x₅ — процент населения с низким статусом.

```python
data = [(0.00632, 6.575, 65.2, 296.0, 4.98),  
(0.02731, 6.421, 78.9, 242.0, 9.14),  
(0.02729, 7.185, 61.1, 242.0, 4.03),  
(0.03237, 6.998, 45.8, 222.0, 2.94),  
(0.06905, 7.147, 54.2, 222.0, 5.33),  
(0.02985, 6.43, 58.7, 222.0, 5.21),  
(0.08829, 6.012, 66.6, 311.0, 12.43)]
```

Добавим в наш набор данных новый признак, который будет равен произведению трёх признаков — x₁, x₄ и x₅.

В результате выполнения программы у вас должен получиться список кортежей. Каждый кортеж должен состоять из шести элементов: первые пять — исходные признаки, а последний, шестой элемент — сгенерированный признак, округлённый до второго знака после запятой.

Результирующий список кортежей занесите в переменную updated_data.

```python
lam = lambda x:(*x, round((x[0]*x[3]*x[4]),2))
updated_data = list(map(lam,data))
print(updated_data)
```

Вы увидели, что использование map() вместо цикла сокращает написание кода вплоть до одной строки. В то же время map() может выполняться даже немного быстрее обычных циклов.

В общем случае map() можно использовать, когда у вас есть набор данных (в виде итерируемого объекта) и вам необходимо совершить над каждым его элементом определённое действие: масштабировать изображение, выделить из каждого предложения главное слово, записать объект в отдельный файл и т. д.

Тогда вы можете выделить ваше действие в отдельную функцию и применить её к каждому элементу из набора данных, используя map().
